```python
import typing 
```

```python
from google import genai 
from google.genai import types 
class Translator():

    def __init__(self, model="gemini-3.6-flash", baseLanguage='polish', targetLanguage='english'):
        self.model = model
        self.baseLanguage = baseLanguage
        self.targetLanguage = targetLanguage
        self.client = genai.Client()
        self.config = types.GenerateContentConfig(
            system_instruction= (
                f"You are a translation agent. Translate each word in the user's input "
                f"from {baseLanguage} into {targetLanguage}. "
                f"Respond with only the translation, no explanations or extra text."
            ),
            temperature=0,
        )
    def translate(self, word):
        response = self.client.models.generate_content(
            model=self.model,
            contents=word,
            config=self.config,
        )
        return response.text

    def translateListOfWords(self, words):
        response = self.client.models.generate_content(
            model=self.model, 
            contents=" ".join(words),
            config=self.config
        )
        return response.text
        

agent = Translator(baseLanguage="polish", targetLanguage="russian")

modelResponse = agent.translateListOfWords(["cześć", "pies", "kot", "zabawka"])
print(modelResponse)
```
