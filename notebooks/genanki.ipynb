```python
import genanki
```

```python
my_model = genanki.Model(
  1607392318,
  'Simple Model',
  fields=[
    {'name': 'Word'},
    {'name': 'Translation'},
    {'name': 'exampleSentence'}
  ],
  templates=[
    {
      'name': 'Word -> Translation',
      'qfmt': '<div class="word">{{Word}}</div>',
      'afmt': '{{FrontSide}}<hr id="answer"><div class="translation">{{Translation}}</div><div class="exampleSentence">{{exampleSentence}}</div>',
    },
  ],
  css="""
      .card {
          font-family: Arial, sans-serif;
          text-align: center;
          background-color: #fafafa;
      }
      .word {
          font-size: 50px;
          color: #1a5fb4;
          font-weight: bold;
      }
      .translation {
          font-size: 40px;
          color: #26a269;
      }
      .exampleSentence {
          font-size: 32px;
          color: #555;
      }
""")

my_note = genanki.Note(
  model=my_model,
  fields=['Pies', 'Dog', 'The dog is a human best friend.'])

my_deck = genanki.Deck(
  2059400111,
  'Animals.2.0')
my_deck.add_note(my_note)

genanki.Package(my_deck).write_to_file('output.apkg')
```
