```python
import genanki
```

```python
my_model = genanki.Model(
  1607392319,
  'Simple Model',
  fields=[
    {'name': 'Question'},
    {'name': 'Answer'},
  ],
  templates=[
    {
      'name': 'Card 1',
      'qfmt': '{{Question}}',
      'afmt': '{{FrontSide}}<hr id="answer">{{Answer}}',
    },
  ])
```

```python
# PYTANIA: 

# (1) czym jest template w modelu? jakie mam moliwości/jak go definiować
# (2) czym są fields? pola jakie sie moga pojawić na karcie (w moim przypadku chcę zeby to było):
#   * słowo
#   * tłumaczenie 
#   * uycie w zdaniu
#   * wymowa
#   * obrazek 
```

```python
my_note = genanki.Note(
  model=my_model,
  fields=['Capital of Argentina', 'Buenos Aires'])
```

```python
print(my_note)
```

```python
my_deck = genanki.Deck(
  2059400110,
  'Country Capitals')

my_deck.add_note(my_note)
```

```python
genanki.Package(my_deck).write_to_file('output.apkg')
```

```python
from google import genai
from pydantic import BaseModel

# 1. Define your data structure
class Recipe(BaseModel):
    recipe_name: str
    ingredients: list[str]

# 2. Initialize the client
client = genai.Client()

# 3. Call generate_content with structured config
response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents="List a popular chocolate chip cookie recipe.",
    config={
        "response_mime_type": "application/json",
        "response_schema": list[Recipe],
    },
)

# 4. Access the structured object
recipes = response.parsed
print(recipes)
```
